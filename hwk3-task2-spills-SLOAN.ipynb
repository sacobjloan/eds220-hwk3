````python
# FINAL CODE

# Import packages
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import geopandas as gpd

# Read in data for NY spills and US Counties
fp = os.path.join('data','Spill_Incidents_20261104.csv')
spills = pd.read_csv(fp)
fp = os.path.join('data','tl_2026_us_county')
counties = gpd.read_file(fp)

# Clean NY spills data
spills.columns = spills.columns.to_series().str.replace(' ','').str.lower() # Clean columns names to not have spaces or any upper case letters
datecol = spills.columns[spills.columns.str.contains('date')].to_list() 
for col in datecol : # Change date columns from character type to datetime type
    spills[col] = pd.to_datetime(spills[col])
spills['quantity'] = spills['quantity'].str.replace(',','') # Clean quantity column and convert to float type
spills['quantity'] = spills['quantity'].astype('float')
petr23 = spills[(spills['spilldate'].dt.year == 2023) & (spills['materialfamily'] == 'Petroleum')] # Select data for petroleum spills in the 2023 calendar year
spills_per_county = petr23.groupby('county')['spillnumber'].nunique().reset_index()
spills_per_county = spills_per_county.rename(columns={'spillnumber':'n_spills'})

# Clean US Counties data
counties.columns = counties.columns.str.lower() # Clean columns names in the US Counties data
ny_counties = counties[counties['statefp'] == '36'] # Select counties in New York State (state code 36)
diff_names_spills = np.setdiff1d(spills_per_county['county'].unique(), ny_counties['name'].unique()) # Find names of counties that are different in the two dfs
spills_per_county.loc[spills_per_county['county'] == 'St Lawrence', 'county'] = 'St. Lawrence' # Correct name of St. Lawrence county to include the period
ny_counties = ny_counties.rename(columns = {'name':'county'}) # Rename county name column

# Merge NY Spill data with US Counties data, then adjust number of spills per county with data from reservations
ny_spills = pd.merge(spills_per_county,ny_counties, on = 'county') # Merge the two cleaned dfs
ny_spills.loc[ny_spills['county'] == 'Onondaga','n_spills'] = ny_spills.loc[ny_spills['county'] == 'Onondaga','n_spills'] + 1 # Add the spill that occured on reservation land
ny_spills.loc[ny_spills['county'] == 'Niagara','n_spills'] = ny_spills.loc[ny_spills['county'] == 'Niagara','n_spills'] + 2 # Add the spills taht occured on reservatin land
ny_spills = gpd.GeoDataFrame(ny_spills)

# Create map of NY Counties by number of chemical spills
fig, ax = plt.subplots()

ax.axis('off')

# Add data to axes
ny_spills.plot(ax=ax,         # Add plot to axes
                 column= 'n_spills',  # Update county colors,
                 legend=True,
                 cmap= 'Reds',
                 )

ax.text(0.5,0,'New York State Department of Environmental Conservation (11/04/26)',
    horizontalalignment= 'center', transform = ax.transAxes, size =9)

# Update axes 
ax.set_title('Number of Spills by County in NY State')

# Display figure
plt.show()


````
